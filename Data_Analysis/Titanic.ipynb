```python
df["Age"] = pd.to_numeric(df["Age"], errors='coerce').astype("Int64")
print(f"Average age: {df['Age'].mean()}")
print(f"Median age: {df['Age'].median()}")
print(f"Minimum Age: {df['Age'].min()}")
print(f"Maximum Age : {df['Age'].max()}")
```

# Passengers Analysis

```python
total_passengers = df.shape[0]
male = df[df['Gender'] == 'Male'].shape[0]
female = df[df['Gender'] == 'Female'].shape[0]

print(f"male : {male}")
print(f"female : {female}")
print(f"total : {total_passengers}")
print(f"male + female : {male+female}")
```

```python
male_percentage = (male / total_passengers) * 100
female_percentage = (female / total_passengers) * 100

print(f"male percentage : {male_percentage:.2f}%")
print(f"female percentage : {female_percentage:.2f}%")
```

```python
youngest_passenger = len(df[df['Age'].between(0, 30)])
oldest_passenger = len(df[df['Age'] > 30])

print(f"youngest passenger : {youngest_passenger}")
print(f"oldest passenger : {oldest_passenger}")
```

```python
average_age = int(df["Age"].mean())
```

```python
child =len(df[df['Age'] < 18])
adult =len(df[df['Age'] >= 18])

print(f"child : {child}")
print(f"adult : {adult}")
```

```python
df['PassengerType'].value_counts()
```

```python
bar = plt.bar(['Male', 'Female','Unknown'], [male, female, 1], color=['blue', 'pink','black'])
plt.xlabel('Gender')
plt.ylabel('Count')
plt.bar_label(bar)
plt.title('Passenger Distribution by Gender')
```

```python
[non_survivor, survivor, unknown] = df['Status'].value_counts()

plt.pie([non_survivor, survivor, unknown], labels=['Non-Survivor', 'Survivor', 'Unknown'], autopct='%1.1f%%', colors=['red', 'blue', 'grey'])
plt.title('Passenger Distribution by Survival Status')
# plt.tight_layout()
plt.show()
```

```python
bar = plt.hist(df['Age'], rwidth=0.85, bins=np.arange(0, 80, 5), color='lightblue', edgecolor='black')
plt.xlabel('Passenger Ages')
plt.ylabel('Number o Passengers')
plt.title('Distribution of Age')
plt.show()
```

```python
male_survival = len(df[(df['Gender'] == 'Male') & (df['Status'] == 'Survivor')])
female_survival = len(df[(df['Gender'] == 'Female') & (df['Status'] == 'Survivor')])

bar = plt.bar(['Male_Survivar', 'Female_Survivor'], [male_survival, female_survival], color=['blue', 'pink'])
plt.xlabel('Gender')
plt.ylabel('Count')
plt.bar_label(bar)
plt.title('Survival Distribution by Gender')
plt.show()
```

```python
survival_counts = df["Status"].value_counts().sort_index()

survival_counts.plot(kind="bar")
plt.xticks([0, 1,2], ["Unknown", "Survivor", "Non-Survivor"], rotation=0)
plt.ylabel("Number of passengers")
plt.title("Titanic Survival Counts")
plt.tight_layout()
plt.show()
```
