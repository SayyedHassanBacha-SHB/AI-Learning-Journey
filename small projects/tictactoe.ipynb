```python

def printStaticInterface():
    print("======"*6)
    print("||\t ||\t  ||\t   ||")
    print("|| [0,0] || [0,1] || [0,2] ||")
    print("||\t ||\t  ||\t   ||")
    print("-----"*7)
    print("||\t ||\t  ||\t   ||")
    print("|| [1,0] || [1,1] || [1,2] ||")
    print("||\t ||\t  ||\t   ||")
    print("-----"*7)
    print("||\t ||\t  ||\t   ||")
    print("|| [2,0] || [2,1] || [2,2] ||")
    print("||\t ||\t  ||\t   ||")
    print("======"*6)
    return

def printDynamicInterface(interface):
    print("======"*6)
    for i in range(3):
        for j in range(3):
            print(f"|| {(interface[i][j]) if ( interface[i][j] ) else (i,j)} ", end="")
        print("||")
        print("-----"*7)
    print("======"*6)
    return
    
def insertValue(pos, value):
    interface[pos[0]][pos[1]] = value
    return

def checkCell(interface, pos):
    if(not(interface[pos[0]][pos[1]])):
        return True
    print("The cell or box is already filled.Pleased chose another cell.")
    return False

def checkUserInput(cellPos):
    inputList = [["0","0"],["0","1"],["0","2"],["1","0"],["1","1"],["1","2"],["2","0"],["2","1"],["2","2"]]
    if(cellPos in inputList):
        return True
    return False
    
    

def winsPattern(interface):
    for i in range(3):
        if(interface[i][0] == interface [i][1] == interface[i][2] != ""):
            return True
        elif(interface[0][i] == interface[1][i] == interface[2][i] != ""):
            return True
    if(interface[0][0] == interface [1][1] == interface[2][2] != ""):
        return True
    elif(interface[0][2] == interface [1][1] == interface[2][0] != ""):
        return True
    else:
        return False

def checkDraw(interface):
    for i in range(3):
        for j in range(3):
            if(interface[i][j] == ""):
                return False
    return True


interface = [["","",""],["","",""],["","",""]]
printStaticInterface()

player1 = input("Enter player1 symbol : ")
player2 = input("Enter player2 symbol : ")

player = True
count = 0

while(True):
    cellPos = list(input(f"{('Player1') if (player) else ('Player2')} Enter cell number that you want in form of list like 01 or 21: "))
    if(not(checkUserInput(cellPos))):
        print("You enter wrong data.Please try again :")
        continue
    cellPos = [int(n) for n in cellPos]
    if(checkCell(interface, cellPos)):
        insertValue(cellPos,player1 if(player) else player2)
        count += 1
    else:
        printDynamicInterface(interface)
        continue
    if(count >= 5 and winsPattern(interface)):
        print(f"{'Player1' if(player) else 'Player2'} wins the match.")
        print("The game ends")
        printDynamicInterface(interface)
        break
    elif(count == 9 and checkDraw(interface)):
        print("The game is over")
        printDynamicInterface(interface)
        break
        
    player = False if player else True
    printDynamicInterface(interface)
    print(count)
# printDynamicInterface(interface)




```
