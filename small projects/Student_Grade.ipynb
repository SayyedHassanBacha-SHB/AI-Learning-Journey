```python
students = [
    {
    "name" : "hassan",
    "age" : 23,
    "marks" : [100,98,90,99,97,100]
    },
    {
    "name" : "hasnain",
    "age" : 27,
    "marks" : [35,45,25,19,38,25]
    },
    {
    "name" : "yaseen",
    "age" : 22,
    "marks" : [88,65,86,75,87,76]
    },
    {
    "name" : "haroon",
    "age" : 19,
    "marks" : [45,67,58,55,67,58]
    },
    {
    "name" : "hamdan",
    "age" : 20,
    "marks" : [86,95,98,83,90,85]
    },
]

def percentage(marks):
    percent = (sum(marks)/600)*100
    return percent

def grade(percent):
    grade = ""
    if(percent >=90 and percent <=100):
        grade = "A"
    elif(percent >= 80 and percent <90):
        grade = "B"
    elif(percent >= 60 and percent < 80):
        grade = "C"
    elif(percent >=40 and percent < 60):
        grade = "D"
    else:
        grade = "F"
    return grade

for student in students:
    percent = percentage(student['marks'])
    student["percentage"] = percent
    grade_st = grade(percent)
    student['grade'] = grade_st

```

```python
import matplotlib.pyplot as plt

percentList = []
gradeList = []
for student in students:
    percentList.append(student['percentage'])
    gradeList.append(student['grade'])

# figure,axes = plt.subplots(2,3)
# plt.scatter(percentList,gradeList)
# plt.bar(percentList,gradeList)
# plt.plot(percentList,gradeList)
plt.pie(percentList, labels= gradeList, autopct="%1.1f")



plt.title("Student DATA",c="red",fontsize="22", fontweight="700")
plt.xlabel("Percentage",c="blue",fontsize="20", fontweight="400")
plt.ylabel("Grade", c="blue",fontsize="20", fontweight="400")
plt.grid()
```
