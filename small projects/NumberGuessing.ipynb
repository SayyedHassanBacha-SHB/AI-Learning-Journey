```python
import random

print("Number Guessing Game")
print("====" * 8)

num = random.randint(1,100)
print(f"num = {num}")
low = 1
high = 100
count = 1

while True :
    guessNum = int((low + high ) / 2)
    print(f"Guess number is : {guessNum}")
    userinput = input("Tell guess number is too high or too low or correct : ")
    if(userinput == "too high" or userinput == "Too high" or userinput == "high"):
        high = guessNum
    elif(userinput == "too low" or userinput == "Too low" or userinput == "low"):
        low = guessNum
    elif(userinput == "correct" or userinput == "Correct"):
        break
    count = count + 1

print(f"We guess the number in {count} attempts")
```
